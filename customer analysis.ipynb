```python
import pandas as pd 
```

```python
df=pd.read_csv('D:\projects\customer churn analysis\datachurn.csv')
df.info()
```

```python
#for fixing total charges to numeric
df['TotalCharges']=pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
#drop the missing values
df=df.dropna()
```

```python
#converting churn to binary 
df['Churn']=df['Churn'].map({'Yes':1, 'No':0})
```

```python
#overall churn rate
df['Churn'].mean()*100
```

```python
#churn By contract
df.groupby('Contract')['Churn'].mean().sort_values(ascending=False)
```

```python
#churn by payment method
df.groupby('PaymentMethod')['Churn'].mean().sort_values(ascending=False)
```

```python
#churn vs tenure
df.groupby('tenure')['Churn'].mean()
#new customers churn more
```

```python
#churn vs montly charges
df.groupby(pd.cut(df['MonthlyCharges'], bins=5))['Churn'].mean()
```

```python
#revenue at risk 
df[df['Churn']==1]['MonthlyCharges'].mean()
```

```python
#creating high risk segments
df.groupby(['Contract','PaymentMethod'])['Churn'].mean().sort_values(ascending=False)
#this gives exact customer groups to target
```
